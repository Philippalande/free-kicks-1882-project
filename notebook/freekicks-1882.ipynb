# Free Kicks: Then & Now
# Using Human-AI Collaboration to Bridge 144 Years of Glasgow Football Culture

Free Kicks at Football is a printed pamphlet of satirical verse and illustration, capturing the humour, rivalry and raw energy of Glasgow football in 1882. Held in Special Collections at the University of Glasgow Library, it was authored by pseudonymous writers "Benedict" and Saunders Wylie, with illustrations by "Jingo." One hundred and forty-three years separate that pamphlet from today yet the culture, the patter, the passion and the tribalism of Glasgow football have never left. This project explores that continuity through an interactive digital presentation, pairing original 1882 verses with AI-generated 2026 equivalents, achieved through precise Human-AI collaboration, generative AI directed and curated by human creative judgment at every stage. Built with modern programming languages, UX techniques and UI frameworks, the project is designed to make its primary audience, Glaswegian football fans, realise they share something profound with the past. Its wider audience of Glasgow citizens, elders, archivists and educators will find in it a new way to access and feel archive material that might otherwise remain unseen. 

## The Archive Object 

The pamphlet is a small, hardcover printed book, its pages rendered in black and white on sand-coloured paper that speaks immediately to its age. The illustrations are hand-drawn in a comic style: crude, expressive and unmistakably Scottish in their humour. The verses match that energy: witty, brutal and sharply satirical, written in a vernacular that feels both of its time and startlingly familiar. Holding it in Special Collections, it felt fragile and precious, a small object carrying an outsized cultural weight, evidence that Glasgow's football culture was already fully formed and fiercely alive in 1882.


## Project Objectives 

1. To introduce the archive object to a general audience so that they realise that the culture, humour, personalities, atmospheres and vibes of Glasgow football are exactly the same today as they were in 1882 and that they share that with the past.
2. To bridge 144 years of Glasgow football culture by using AI to generate contemporary equivalents of the pamphlet's humour, demonstrating cultural continuity through creative transformation.
3. To demonstrate that Human-AI collaboration can synthesise the stylistic and thematic DNA of the 1882 verse: the satirical tone, the first-person voice, the physicality, the dark humour, and reproject it into 2026. A human alone would write a pastiche. AI alone would produce something generic. Together they produce something genuinely neither.


## The Three Spreads 

- Spread 1 — A Lay-On: The Leather was selected because the brutal, physical language of the football speaks directly to the same raw culture Glasgow football fans embody today. The imagery of being battered, kicked and fought over is as recognisable in 2026 as it was in 1882.
- Spread 2 — Half-Time was selected because the double meaning of "dribbling" and "passing" (simultaneously referencing football and drinking) captures the patter and dark humour that remains central to Glasgow football culture. The half-time drink is as much a ritual today as it was 144 years ago.
- Spread 3 — England versus Scotland, 1882 was selected because the tribal rivalry and crowd energy depicted at Hampden Park mirrors the same passion seen today in fixtures like Celtic versus Rangers. The national pride, the chaos and the collective roar of the crowd are unchanged across generations.

## Methodology 

This project employs a structured Human-AI collaborative methodology across three spreads. For each spread, the original 1882 verse was first analysed by myself (the human collaborator) to identify its core themes, satirical tone, vernacular voice and cultural context. This analysis informed the design of a precise natural language prompt, which was submitted to OpenAI's GPT model via the OpenAI API. The AI generated a contemporary 2026 equivalent verse, which was then evaluated by the human collaborator accepted, rejected or iteratively refined based on whether it authentically captured the cultural continuity between 1882 and 2026. Every curatorial decision is documented within this notebook. The final outputs represent neither pure human creativity nor pure AI generation, but a genuinely collaborative product where human judgment directs and curates AI capability at every stage.

### Primary Vernacular Research
A key methodological decision was the collection of authentic Glasgow football vernacular as a primary research dataset to inform prompt engineering. Rather than relying solely on GPT's training data to supply culturally accurate language, contemporary supporter comments were sourced directly from the public Instagram posts of Celtic FC and Rangers FC in May 2026. This dataset provided real, unfiltered Glasgow football patter from both sides of the city's football divide, including phrases such as "mon the bhoys", "into them", "dog pish", "by any means necessary" and "you won't be here by Wednesday." These phrases were embedded directly into the prompts for each spread, grounding the AI outputs in authentic contemporary vernacular rather than generalised language patterns. This decision was entirely human in origin and proved decisive in elevating the cultural authenticity of the generated verses. It also reflects a broader methodological principle of this project — that the human collaborator must actively supply cultural context that the model cannot be assumed to hold independently.

### Understanding the Generative AI Tool 

The generative AI tool employed in this project is OpenAI's GPT, a Large Language Model (LLM) built on the Transformer architecture. As Foster (2019) establishes, GPT belongs to the autoregressive family of generative models, predicting each subsequent token based on all preceding ones, generating language sequentially through learned probability distributions rather than fixed rules. Following the generative ML workflow introduced in this course, the training data is a large corpus of human written text including books, articles and published literature; the ML algorithm is the Transformer; the model is GPT itself; and the new data, the prediction result, is the contemporary verse generated in response to each human designed prompt. It is important to note that OpenAI has not fully disclosed the composition of GPT's training data, and therefore no absolute claim can be made regarding its specific knowledge of Scottish vernacular or Victorian satirical verse. This uncertainty is precisely why human curatorial judgment is indispensable at every stage of this project. The methodology therefore employs a process of deliberate prompt engineering, in which each prompt is carefully constructed to supply the cultural, linguistic and tonal context the model requires, rather than assuming the model holds that knowledge independently. For each spread, multiple prompt iterations were tested, with outputs evaluated against specific human criteria: authenticity of Glasgow vernacular, satirical tone, cultural continuity between 1882 and 2026, and emotional resonance with the target audience. Outputs that failed these criteria were rejected or refined iteratively until the human collaborator was satisfied. This positions the Human-AI collaboration not as a delegation of creative responsibility to the model, but as a structured creative dialogue in which the human directs, evaluates and curates at every stage.

```python
import os 
from dotenv import load_dotenv 
from openai import OpenAI

load_dotenv()

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
```

### Spread 1 — A Lay-On: The Leather

Spread 1 draws from the opening verse of the pamphlet, A Lay-On: The Leather, in which the football itself speaks in first person, describing the physical brutality of the game from its own perspective. The tone is visceral, darkly comic and relentless, capturing the raw physicality of 1882 Glasgow football through the voice of an object that never gets a moment's rest. This voice is timeless — the battered football, the endless kicks, the chaos of the crowd — these images resonate just as powerfully with a 2026 Glasgow football fan watching a match at Hampden or Ibrox. The AI was directed to generate a 2026 equivalent verse maintaining the first person voice of the football, the satirical tone, the Glasgow vernacular and the same relentless rhythmic energy, transposed into the contemporary game.

```python
# Define the original 1882 verse as a variable so we can reference it clearly
verse_1 = """
I'm batter'd, batter'd to and fro,
At rest a moment never,
For men may come and men may go,
But I go on for ever.
I'm bang'd about, now in now out,
Now with a dribbler sailing,
There's here and there a lusty shout,
And here and there a wailing.
By thirty kicks I'm hurried down,
Or slipp'd beyond the edges,
By twenty fouls, a broken crown,
And half a hundred dodges.
"""

# Make the API call to GPT
# client.chat.completions.create sends our prompt to the model
response = client.chat.completions.create(
    model="gpt-4o-mini",  # cost effective GPT model
    messages=[
        {
            "role": "system",  # tells GPT how to behave
            "content": "You are an expert in Scottish vernacular poetry and satirical verse. You have deep knowledge of Glasgow football culture, both historical and contemporary."
        },
        {
            "role": "user",  # our actual prompt
            "content": f"""Here is an 1882 satirical verse from a Glasgow pamphlet called Free Kicks at Football. 
            The football speaks in first person about being kicked around during a match.
            
            Original verse:
            {verse_1}
            
            Please generate a 2026 equivalent verse that:
            - Keeps the first person voice of the football
            - Maintains the satirical, darkly comic tone
            - Uses contemporary Glasgow football vernacular and references
            - Matches the rhythmic structure of the original
            - References modern elements like VAR, social media, stadium atmosphere
            - Feels authentically Glaswegian"""
        }
    ]
)

# Extract and print the generated verse from the response
generated_verse_1 = response.choices[0].message.content
print(generated_verse_1)
```

The initial output was rejected as the vernacular did not authentically reflect Glasgow football culture. The prompt was refined by injecting real Glasgow football language sourced directly from Celtic and Rangers Instagram comments in May 2026, used here as a primary research dataset to ensure cultural accuracy

```python
# Refined prompt incorporating authentic Glasgow football vernacular
# sourced from Celtic and Rangers Instagram comments, May 2026

response_2 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "You are an expert in Scottish vernacular poetry and Glasgow football culture. You write in authentic Glaswegian dialect."
        },
        {
            "role": "user",
            "content": f"""Here is an 1882 satirical verse from a Glasgow pamphlet called Free Kicks at Football.
The football speaks in first person about being kicked around during a match.

Original verse:
{verse_1}

Generate a 2026 equivalent verse that:
- Keeps the first person voice of the football
- Uses AUTHENTIC Glasgow football vernacular from both Celtic and Rangers fans such as: 'mon the bhoys', 'into them', 'get the job done', 'by any means necessary', 'HH', 'dog pish', 'bunch of clowns', 'rohl out', 'utter embarrassment', 'out of his depth', 'midfield got bossed', 'no game plan', 'time to go', 'cheers for the points lads'
- Maintains the satirical darkly comic tone of the original
- Matches the rhythmic structure and stanza length of the original
- References modern elements like social media, stadium atmosphere
- Feels raw, funny and authentically Glaswegian
- Avoids English football slang like 'tekkers' or 'lads'"""
        }
    ]
)

generated_verse_1_refined = response_2.choices[0].message.content
print(generated_verse_1_refined)
```

The refined output was accepted. The second iteration successfully incorporated authentic Glasgow football vernacular sourced from real supporter comments, including references to "mon the bhoys", "dog pish", "bunch o' clowns" and "HH"(Hair Hail). The satirical tone, first person voice and rhythmic structure of the original 1882 verse were maintained. Minor inconsistencies in phrasing were noted but did not detract from the overall cultural authenticity and tonal accuracy of the output. This verse will be presented alongside the original 1882 verse in Spread 1 of the interactive digital presentation, representing the 2026 equivalent generated through Human-AI collaboration.

### Spread 2 — Half-Time

Spread 2 draws from the illustration and caption titled Half-Time, in which two men share a drink at the interval, one remarking: "Haud on for goodness sake man: ye're vera guid at dribbling but a puir han at passing." The humour operates on a double meaning — dribbling and passing refer simultaneously to football skills and the act of drinking. This layered vernacular wit is quintessentially Glaswegian and remains just as recognisable in 2026. The AI was directed to generate a 2026 equivalent that maintains the double meaning, the half-time setting, the sardonic wit and the authentic Glasgow voice.

```python
# Spread 2 — Half-Time
# Original caption from the 1882 illustration

verse_2 = """
Half-Time
Haud on for goodness sake man: ye're vera guid at dribbling 
but a puir han at passing!
"""

response_3 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "You are an expert in Scottish vernacular poetry and Glasgow football culture. You write in authentic Glaswegian dialect."
        },
        {
            "role": "user",
            "content": f"""Here is an 1882 satirical illustration caption from a Glasgow pamphlet called Free Kicks at Football.
Two men share a drink at half-time. The humour works on a double meaning where dribbling and passing refer to both football and drinking.

Original caption:
{verse_2}

Generate a 2026 equivalent short verse or dialogue that:
- Maintains the half-time setting
- Keeps the double meaning of football skills and drinking
- Uses authentic Glasgow football vernacular such as: mon the bhoys, into them, pure dead brilliant, by any means necessary, dog pish, HH, COYBIG
- Is sardonic, witty and authentically Glaswegian
- References modern half-time culture such as phones, pies, VAR debates, social media
- Feels like genuine Glasgow pub or stadium banter"""
        }
    ]
)

generated_verse_2 = response_3.choices[0].message.content
print(generated_verse_2)
```

The initial Spread 2 output was rejected. While the verse incorporated authentic Glasgow vernacular and the half-time setting, several lines were nonsensical and broke the internal logic of the piece. Phrases such as "more like a shot o' ghost" and "I'm a witch" had no cultural grounding and undermined the satirical tone. The prompt was refined to request tighter logical coherence and a shorter, punchier format closer to the original caption style.

```python
# Spread 2 — Half-Time — Refined Prompt v2

response_5 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "You are an expert in Scottish vernacular humour and Glasgow football culture. You write sharp, witty poetry in authentic Glaswegian dialect."
        },
        {
            "role": "user",
            "content": f"""Here is an 1882 satirical illustration caption from a Glasgow pamphlet called Free Kicks at Football.
Two men share a drink at half-time. The humour works on a double meaning where dribbling and passing refer to both football skills and drinking alcohol.

Original caption:
{verse_2}

Generate a 2026 equivalent that:
- Is a single stanza of six to eight lines
- Is a witty remark or short dialogue between two fans at half-time
- Maintains the SAME double meaning — dribbling and passing referring to both football and drinking
- Uses authentic Glasgow vernacular naturally — do not force in too many slang terms
- Is logically coherent — every line must make sense
- Is sardonic and funny in a distinctly Glaswegian way
- References modern half-time culture naturally — pies, phones, a pint
- Avoids nonsensical metaphors"""
        }
    ]
)

generated_verse_2_refined = response_5.choices[0].message.content
print(generated_verse_2_refined)
```

The second Spread 2 output was rejected. While significantly improved from the first iteration, two terms were identified as inauthentic to Glasgow vernacular. "Mate" and "lad" are more commonly associated with English football culture rather than Glaswegian speech, and the word "monochrome" felt tonally out of place in an otherwise colloquial verse. 

```python
# Spread 2 — Half-Time — Refined Prompt v3

response_6 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "You are an expert in Scottish vernacular humour and Glasgow football culture. You write sharp, witty poetry in authentic Glaswegian dialect."
        },
        {
            "role": "user",
            "content": f"""Here is an 1882 satirical illustration caption from a Glasgow pamphlet called Free Kicks at Football.
Two men share a drink at half-time. The humour works on a double meaning where dribbling and passing refer to both football skills and drinking alcohol.

Original caption:
{verse_2}

Generate a 2026 equivalent that:
- Is a single stanza of six to eight lines
- Maintains the SAME double meaning of dribbling and passing referring to both football and drinking
- Uses authentic Glasgow vernacular — use 'pal' or 'son' instead of 'mate' or 'lad'
- Is logically coherent — every line must make sense
- Avoids overly formal or out of place words like 'monochrome'
- Is sardonic and funny in a distinctly Glaswegian way
- References modern half-time culture naturally — pies, phones, a pint
- Avoids nonsensical metaphors"""
        }
    ]
)

generated_verse_2_refined2 = response_6.choices[0].message.content
print(generated_verse_2_refined2)
```

After multiple iterations, the version 3 output was accepted as the final result for Spread 2. While minor phrasing awkwardness was noted in phrases such as "skill o' a wee nibble", subsequent iterations produced less coherent results overall, demonstrating a key limitation of autoregressive generative models — further iteration does not guarantee improved output quality. An attempt was made to manually edit the verse however the constraints of the rhyme scheme made this unfeasible without significantly altering the structure. This reinforces the importance of human curatorial judgment in identifying the optimal output across multiple iterations, rather than defaulting to the most recent generation. The accepted verse successfully maintains the double meaning of the original 1882 caption, uses authentic Glasgow vernacular and captures the sardonic half-time humour that connects 1882 and 2026.


### Spread 3 — England versus Scotland, 1882
Spread 3 draws from the verse titled England versus Scotland, 1882 — A Reminiscence of Hampden Park. The verse captures the chaos, tribalism and raw crowd energy of an international match at Hampden, with fans storming the ropes, police overwhelmed and national pride at fever pitch. This theme is timelessly Glaswegian — the Scotland versus England rivalry remains one of the most emotionally charged fixtures in world football, and the chaos of the crowd at Hampden is as recognisable in 2026 as it was in 1882. The AI was directed to generate a 2026 equivalent verse maintaining the crowd energy, the national rivalry, the chaos and the Glaswegian voice, transposed into the contemporary game with modern references.

```python
# Spread 3 — Refined Prompt v3 — Tighter structure

response_10 = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "You are a Glaswegian poet who writes short punchy satirical verse about football in authentic Scottish dialect."
        },
        {
            "role": "user",
            "content": f"""Write a short punchy verse about Scotland fans at Hampden in 2026 facing England.

Rules:
- Maximum 3 stanzas of 4 lines each
- Every line must make complete sense
- Use only: aye, nae, pure, cannae, mon the bhoys, into them, son, pal, by any means
- Reference iPhones and Hampden atmosphere
- Raw passionate and funny
- No nonsense lines"""
        }
    ]
)

generated_verse_3_refined2 = response_10.choices[0].message.content
print(generated_verse_3_refined2)
```

The third Spread 3 output was accepted. A simpler, tighter prompt structure produced the strongest result — demonstrating that over-engineering prompts can reduce output quality. The verse successfully captures the raw crowd energy, national pride and Glaswegian atmosphere of Hampden in 2026, directly echoing the chaos and tribalism of the 1882 original. Authentic vernacular including "mon the bhoys", "cannae", "pal" and "son" ground the verse culturally, while the reference to iPhones and floodlights places it firmly in the contemporary moment. This output required three iterations, reinforcing the central methodology argument that Human-AI collaboration is an iterative creative dialogue rather than a single generation event.
